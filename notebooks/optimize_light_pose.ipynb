```python
import os
import json
import numpy as np
from pathlib import Path
import plotly.graph_objects as go
```

```python
data_path = Path(
    os.path.expanduser("/home/shared/data_nerfstudio/ShadowSplat/blender/perseverance/")
)
transforms_path = data_path / "transforms.json"
image_path = data_path / "images_split"
with open(transforms_path, "r") as f:
    transforms = json.load(f)
```

```python
positions = [np.asarray(frame["transform_matrix"])[:3, 3] for frame in transforms["frames"]]
positions = np.array(positions)
```

```python
positions_subset = []
for pos in positions:
    if pos[0] > 2.0:
        positions_subset.append(pos)
positions_subset = np.array(positions_subset)
fig = go.Figure(
    data=go.Scatter3d(
        x=positions[:, 0],
        y=positions[:, 1],
        z=positions[:, 2],
        mode="markers",
        marker=dict(color="blue", size=5),
    )
)
fig.add_trace(
    go.Scatter3d(
        x=positions_subset[:, 0],
        y=positions_subset[:, 1],
        z=positions_subset[:, 2],
        mode="markers",
        marker=dict(color="red", size=5),
    )
)
fig.update_layout(height=800, width=1200, scene=dict(aspectmode="data"))
fig.show()
```

```python
transforms_subset = transforms.copy()
subset_frames = []
for frame in transforms_subset["frames"]:
    if np.asarray(frame["transform_matrix"])[:3, 3][0] > 2:
        subset_frames.append(frame)
    else:
        # Add with probablity 0.1
        if np.random.rand() < 0.1:
            subset_frames.append(frame)
transforms_subset["frames"] = subset_frames
```

```python
with open(data_path / "transforms_subset.json", "w") as f:
    json.dump(transforms_subset, f, indent=4)
```
