```python
import numpy as np
import open3d as o3d
import matplotlib.pyplot as plt
```

```python
# mesh = o3d.io.read_triangle_mesh("../data/models/master_chief/master_chief.obj")
# mesh = o3d.io.read_triangle_mesh("../data/models/penguin/PenguinBaseMesh.obj")
mesh = o3d.io.read_triangle_mesh("../data/models/bunny.obj")
mesh.compute_vertex_normals()

print(mesh.has_textures())  # True
print(mesh.has_triangle_uvs())  # True

# o3d.visualization.draw_geometries([mesh])
```

```python
mesh.paint_uniform_color([0.7, 0.7, 0.7])
```

```python
# Setup offscreen renderer
w, h = 1024, 1024
renderer = o3d.visualization.rendering.OffscreenRenderer(w, h)
scene = renderer.scene

# Configure material (use default if mesh has textures)
material = o3d.visualization.rendering.MaterialRecord()
material.shader = "defaultLit"

scene.clear_geometry()
scene.add_geometry("mesh", mesh, material)

# Setup camera (front view)
bbox = mesh.get_axis_aligned_bounding_box()
center = bbox.get_center()
extent = bbox.get_extent().max()
cam_pos = center + np.array([0, 0, extent * 1.0])  # front-on, +Z
up = [0, 1, 0]  # Y up
scene.camera.look_at(center, cam_pos, up)

# Add lighting
scene.set_lighting(scene.LightingProfile.SOFT_SHADOWS, [0, 0, 0])

# Render image
image = renderer.render_to_image()
# o3d.io.write_image("rendered_front_view.png", image)
# Display the image
plt.imshow(image)
plt.axis("off")
plt.show()
```
